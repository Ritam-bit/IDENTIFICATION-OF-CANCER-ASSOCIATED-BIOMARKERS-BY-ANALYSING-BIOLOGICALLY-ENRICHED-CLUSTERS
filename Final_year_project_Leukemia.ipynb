```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from numpy.linalg import inv

# Load the preprocessed dataset
file_path = "Normalized_Lukemia_Gene_Expression_Normal.xlsx"
df = pd.read_excel(file_path, header=[0, 1,2])

# Separate metadata and numeric columns
metadata_cols = df.iloc[:, :2]   # ClusterID, Sample Name, etc.
numeric_data = df.iloc[:, 3:]    # Only numeric values for clustering
X = numeric_data.to_numpy()
```

```python
def initialize_membership_matrix(n_samples, n_clusters):
    membership_mat = np.random.random((n_samples, n_clusters))
    membership_mat /= np.sum(membership_mat, axis=1, keepdims=True)
    return membership_mat

def calculate_cluster_center(X, membership_mat, m):
    n_clusters = membership_mat.shape[1]
    centers = np.zeros((n_clusters, X.shape[1]))
    for i in range(n_clusters):
        num = np.sum((membership_mat[:, i, np.newaxis] ** m) * X, axis=0)
        den = np.sum(membership_mat[:, i] ** m)
        centers[i, :] = num / den
    return centers

def compute_covariance_matrix(X, centers, membership_mat, m):
    n_clusters = centers.shape[0]
    cov_matrices = [np.zeros((X.shape[1], X.shape[1])) for _ in range(n_clusters)]
    for j in range(n_clusters):
        num = np.zeros((X.shape[1], X.shape[1]))
        den = np.sum(membership_mat[:, j] ** m)
        for i in range(X.shape[0]):
            diff = (X[i, :] - centers[j, :])[np.newaxis]
            num += (membership_mat[i, j] ** m) * np.dot(diff.T, diff)
        cov_matrices[j] = num / den
    return cov_matrices

def calculate_distance(X, centers, cov_matrices):
    n_samples = X.shape[0]
    n_clusters = centers.shape[0]
    distances = np.zeros((n_samples, n_clusters))
    for i in range(n_samples):
        for j in range(n_clusters):
            diff = (X[i, :] - centers[j, :])[np.newaxis]
            inv_cov = inv(cov_matrices[j])  # Inverse covariance
            distances[i, j] = np.sqrt(np.dot(np.dot(diff, inv_cov), diff.T))
    return distances

def update_membership_matrix(X, centers, m):
    n_samples = X.shape[0]
    n_clusters = centers.shape[0]
    new_membership_mat = np.zeros((n_samples, n_clusters))
    for i in range(n_samples):
        for j in range(n_clusters):
            sum_term = 0
            for k in range(n_clusters):
                num = np.linalg.norm(X[i, :] - centers[j, :])
                den = np.linalg.norm(X[i, :] - centers[k, :])
                sum_term += (num / den) ** (2 / (m - 1))
            new_membership_mat[i, j] = 1 / sum_term
    return new_membership_mat

def gustafson_kessel(X, n_clusters, max_iter=100, m=2.0):
    membership_mat = initialize_membership_matrix(X.shape[0], n_clusters)
    for _ in range(max_iter):
        centers = calculate_cluster_center(X, membership_mat, m)
        cov_matrices = compute_covariance_matrix(X, centers, membership_mat, m)
        distances = calculate_distance(X, centers, cov_matrices)
        membership_mat = update_membership_matrix(X, centers, m)
    return centers, membership_mat
```

```python
n_clusters = 3  # You can modify this based on your dataset
centers, membership_mat = gustafson_kessel(X, n_clusters)

# Assign clusters based on highest membership score
df["GK_Cluster"] = membership_mat.argmax(axis=1)

# Reset the column index to a single level before saving
df.columns = df.columns.get_level_values(0)  # Or df.columns = range(len(df.columns))

# Save the clustered dataset
df.to_excel("GK_Clustered_Lukemia_Gene_Expression_Normal.xlsx", index=False, engine='openpyxl')

print("Clustering completed! Clustered dataset saved.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from numpy.linalg import inv

# Load the preprocessed dataset
file_path = "Normalized_Lukemia_Gene_Expression_Normal.xlsx"
df = pd.read_excel(file_path, header=[0, 1,2])

# Separate metadata and numeric columns
metadata_cols = df.iloc[:, :2]   # ClusterID, Sample Name, etc.
numeric_data = df.iloc[:, 3:]    # Only numeric values for clustering
X = numeric_data.to_numpy()
```

```python
import numpy as np
from tqdm import tqdm
from scipy.spatial.distance import cdist
import matplotlib.pyplot as plt

# --- VALIDITY FUNCTIONS ---

def xie_beni_index(X, centers, membership_mat, m):
    num = np.sum((membership_mat ** m) * (cdist(X, centers) ** 2))
    inter_cluster_dist = cdist(centers, centers)
    np.fill_diagonal(inter_cluster_dist, np.inf)
    min_dist = np.min(inter_cluster_dist)
    return num / (X.shape[0] * min_dist ** 2)

def fukuyama_sugeno_index(X, centers, membership_mat, m):
    global_center = np.mean(X, axis=0)
    first_term = np.sum((membership_mat ** m) * (cdist(X, centers) ** 2))
    second_term = np.sum(np.linalg.norm(centers - global_center, axis=1) ** 2)
    return first_term - second_term
```

```python
def batched_validity_indices(X, k_range=range(2, 21), batch_size=5000, m=2.0):
    xb_all = {k: [] for k in k_range}
    fs_all = {k: [] for k in k_range}

    for start in tqdm(range(0, X.shape[0], batch_size), desc="Normal Batches"):
        end = min(start + batch_size, X.shape[0])
        batch = X[start:end]

        for k in k_range:
            centers, membership = gustafson_kessel(batch, n_clusters=k, max_iter=50, m=m)
            xb_val = xie_beni_index(batch, centers, membership, m)
            fs_val = fukuyama_sugeno_index(batch, centers, membership, m)

            xb_all[k].append(xb_val)
            fs_all[k].append(fs_val)

    xb_avg = [np.mean(xb_all[k]) for k in k_range]
    fs_avg = [np.mean(fs_all[k]) for k in k_range]

    return k_range, xb_avg, fs_avg
```

```python
k_vals, xb_normal, fs_normal = batched_validity_indices(
    X, k_range=range(2, 21), batch_size=5000, m=2.0
)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the membership score datasets
file_path = "GK_Membership_Scores (3).xlsx"
df_normal = pd.read_excel(file_path, sheet_name="Normal")
df_carcinogenic = pd.read_excel(file_path, sheet_name="Carcinogenic")

# Extract only the numeric membership scores (excluding metadata columns)
membership_cols_normal = df_normal.columns[2:]  # Assuming first 3 columns are metadata
membership_cols_carcinogenic = df_carcinogenic.columns[2:]

normal_scores = df_normal[membership_cols_normal].values  # Shape (num_genes, 3)
carcinogenic_scores = df_carcinogenic[membership_cols_carcinogenic].values  # Shape (num_genes, 14)

# **Step 1: Compute Max Membership for Each Gene**
max_normal = np.max(normal_scores, axis=1)  # Shape (num_genes,)
max_carcinogenic = np.max(carcinogenic_scores, axis=1)  # Shape (num_genes,)

# **Step 2: Compute Absolute Difference (Delta U)**
delta_u = np.abs(max_carcinogenic - max_normal)

# **Step 3: Set Dynamic Threshold**
mean_delta_u = np.mean(delta_u)
std_delta_u = np.std(delta_u)
threshold = mean_delta_u + 2 * std_delta_u  # Mean + 2*Std Dev

# **Step 4: Identify Significant Genes**
significant_gene_indices = np.where(delta_u < threshold)[0]
significant_genes = df_normal.iloc[significant_gene_indices, :2]  # Keeping metadata columns

# Save the significant genes
output_path = "Significant_Genes.xlsx"
significant_genes.to_excel(output_path, index=False, engine='openpyxl')

# **Step 5: Visualize Delta U Distribution**
plt.figure(figsize=(10, 6))
sns.histplot(delta_u, bins=50, kde=True, color="blue")
plt.axvline(threshold, color="red", linestyle="dashed", label=f"Threshold ({threshold:.4f})")
plt.title("Membership Score Differences (ΔU) - Dynamic Thresholding")
plt.xlabel("ΔU Value")
plt.ylabel("Frequency")
plt.legend()
plt.show()

print(f"Total significant genes identified: {len(significant_genes)}")
print(significant_genes.head())
```

```python
import pandas as pd

# --- Step 1: Load Files ---

# Upload your files in Colab first if not already:
# from google.colab import files
# uploaded = files.upload()

# Load predicted genes
predicted_genes_df = pd.read_excel("Significant_Genes (3).xlsx")
predicted_genes = predicted_genes_df.iloc[:, 0].dropna().str.upper().tolist()

# Load reference file (Symbol + Aliases)
reference_df = pd.read_excel("vnd.openxmlformats-officedocument.spreadsheetml.sheet&rendition=1 (1).xlsx")

# --- Step 2: Prepare the Symbol + Aliases Mapping ---

# Create a set of all known names (symbols + aliases)
reference_names_set = set()

for idx, row in reference_df.iterrows():
    symbol = str(row['Symbol']).strip().upper()
    aliases = str(row['Aliases']).strip().upper().replace(';', ',').replace('|', ',')  # Handle different separators
    aliases_list = [alias.strip() for alias in aliases.split(',')] if aliases else []

    # Add both the symbol and aliases
    reference_names_set.add(symbol)
    reference_names_set.update(aliases_list)

# Remove any empty strings
reference_names_set = set(name for name in reference_names_set if name)

# --- Step 3: Match Predicted Genes with Reference ---

matched_genes = [gene for gene in predicted_genes if gene in reference_names_set]
unmatched_genes = [gene for gene in predicted_genes if gene not in reference_names_set]

# --- Step 4: Metrics Calculation ---

# True Positives (TP): matched genes
TP = len(matched_genes)

# False Positives (FP): predicted but not actually in known list
FP = len(unmatched_genes)

# False Negatives (FN): known genes not predicted
FN = len(reference_names_set - set(predicted_genes))

# Total Predicted = 291
Total_Predicted = len(predicted_genes)

# Accuracy = TP / Total Predicted
accuracy = TP / Total_Predicted

# --- Step 5: Display Results ---

print(f"✅ Total Genes in Reference List (Symbol + Aliases): {len(reference_names_set)}")
print(f"✅ Total Predicted Genes: {Total_Predicted}")
print(f"✅ Total Matched Genes: {TP}")
print(f"✅ Total Unmatched Genes: {FP}")
print("")

print(f"🔵 Accuracy: {accuracy:.3f}")
```
